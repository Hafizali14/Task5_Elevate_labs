```python
# Check structure, data types, and non-null counts
print("--- DATASET INFO ---")
df.info()

# Check summary statistics for numerical columns
print("\n--- SUMMARY STATISTICS ---")
display(df.describe())

# Check missing values count
print("\n--- MISSING VALUES ---")
print(df.isnull().sum())
```

```python
print("Survival Counts (0 = Died, 1 = Survived):")
print(df['Survived'].value_counts(), "\n")

print("Passenger Class Counts:")
print(df['Pclass'].value_counts(), "\n")

print("Gender Counts:")
print(df['Sex'].value_counts(), "\n")

print("Embarkation Port Counts:")
print(df['Embarked'].value_counts())
```

### Observations: Statistical Inspection
* **Dataset Size:** The dataset contains 891 rows (passengers) and 12 columns.
* **Missing Values:** `Cabin` has the most missing values (687 missing), followed by `Age` (177 missing) and `Embarked` (2 missing).
* **Survival Rate:** Out of 891 passengers, 549 (~61.6%) did not survive, while 342 (~38.4%) survived.
* **Demographics & Class:** Males make up ~65% of the passengers (577 males vs. 314 females), and more than half of the passengers (491) traveled in 3rd Class.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Age Distribution
sns.histplot(df['Age'].dropna(), bins=30, kde=True, color='steelblue', ax=axes[0])
axes[0].set_title('Age Distribution of Passengers')
axes[0].set_xlabel('Age')

# Fare Distribution
sns.histplot(df['Fare'], bins=40, kde=True, color='teal', ax=axes[1])
axes[1].set_title('Fare Distribution of Passengers')
axes[1].set_xlabel('Fare')

plt.tight_layout()
plt.show()
```

### Observations: Histograms (Age & Fare)
* **Age Distribution:** Roughly bell-shaped (normal) with a slight right skew. Most passengers were young adults between 20 and 40 years old, with a noticeable spike for infants and young children (ages 0–5).
* **Fare Distribution:** Heavily right-skewed. The vast majority of tickets cost under $50 (median fare is ~$14.45), with a long tail extending past $500.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Age Boxplot
sns.boxplot(x=df['Age'], color='skyblue', ax=axes[0])
axes[0].set_title('Boxplot of Passenger Age')

# Fare Boxplot
sns.boxplot(x=df['Fare'], color='lightcoral', ax=axes[1])
axes[1].set_title('Boxplot of Ticket Fare')

plt.tight_layout()
plt.show()
```

### Observations: Boxplots & Anomalies
* **Age Outliers:** The interquartile range (IQR) spans roughly from 20 to 38 years old, with a median of 28. Passengers aged above ~65 appear as upper outliers.
* **Fare Outliers:** Numerous high-value outliers exist above ~$65, including extreme luxury fares exceeding $500 (1st-class suites).

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Survival by Sex
sns.countplot(x='Sex', hue='Survived', data=df, palette='Set2', ax=axes[0])
axes[0].set_title('Survival Count by Gender')
axes[0].legend(title='Survived', labels=['No (0)', 'Yes (1)'])

# Survival by Passenger Class
sns.countplot(x='Pclass', hue='Survived', data=df, palette='Set1', ax=axes[1])
axes[1].set_title('Survival Count by Passenger Class')
axes[1].legend(title='Survived', labels=['No (0)', 'Yes (1)'])

plt.tight_layout()
plt.show()
```

### Observations: Survival by Gender & Class
* **Gender Impact:** Females had a drastically higher survival rate (~74%) compared to males (~19%), reflecting the "women and children first" evacuation protocol.
* **Socioeconomic Class Impact:** 1st-class passengers had the highest survival rate (~63%), followed by 2nd-class (~47%), whereas 3rd-class passengers suffered the highest fatalities (~76% did not survive).

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(x='Age', y='Fare', hue='Survived', style='Sex', data=df, alpha=0.75, palette='coolwarm')
plt.title('Age vs. Fare Colored by Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Observations: Age vs. Fare Scatterplot
* Passengers who paid higher fares (above $100) clustered heavily in the survivor group across almost all age brackets.
* Young children (under age 10) also show a higher concentration of survivors even at lower fare tiers.

```python
selected_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
sns.pairplot(df[selected_cols].dropna(), hue='Survived', palette='husl', diag_kind='kde')
plt.suptitle('Pairplot of Key Numerical Features by Survival', y=1.02)
plt.show()
```

### Observations: Pairplot
* **Class & Fare Separation:** 1st-class passengers clearly separate into higher fare ranges and higher survival density curves.
* **Family Size (`SibSp` & `Parch`):** Solo travelers (`SibSp = 0`, `Parch = 0`) represent the largest group overall, though passengers traveling with small families (1–2 relatives) show slightly better survival ratios than large families (4+ relatives).

```python
plt.figure(figsize=(10, 7))
corr_matrix = df.corr(numeric_only=True)
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5, vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Numerical Features')
plt.tight_layout()
plt.show()
```

### Observations: Correlation Heatmap
* **Strongest Negative Correlation (`Pclass` vs. `Fare`: -0.55):** As class number increases from 1 to 3 (1st to 3rd class), ticket fare drops significantly.
* **Survival Correlations:** `Pclass` has a moderate negative correlation with `Survived` (-0.34), while `Fare` has a positive correlation with `Survived` (+0.26), confirming that wealthier passengers had better survival odds.
* **Family Correlation (`SibSp` vs. `Parch`: +0.41):** Passengers traveling with siblings/spouses frequently also traveled with parents/children.
* **Multicollinearity Check:** No two independent features exceed a correlation of 0.60, indicating low multicollinearity among numerical variables.

## Summary of Findings
1. **Primary Survival Drivers:** Gender and passenger class were the strongest determinants of survival. Female passengers and 1st-class ticket holders survived at substantially higher rates than males and 3rd-class passengers.
2. **Wealth & Fare Skewness:** Ticket fares are heavily right-skewed with extreme upper outliers (over $500). Higher ticket fares positively correlate with survival (+0.26).
3. **Age Trends:** Most passengers were between 20 and 40 years old, though young children had noticeably higher survival rates compared to adult males.
4. **Data Quality & Missingness:** `Cabin` is missing ~77% of its values and should be dropped or converted into a binary `Has_Cabin` indicator before modeling, while `Age` (~20% missing) can be imputed using median age grouped by `Pclass` and `Sex`.
